```python
from datasets import load_dataset
from pathlib import Path
import warnings
warnings.filterwarnings('ignore')
```

```python
import os
```

```python
data=load_dataset(
    path='bijinc/vimeo-90k-mini',
    split='train',
    streaming=True
)

root=Path("dataset/train")
root.mkdir(parents=True,exist_ok=True)

for i,imgs in enumerate(data):
    if i>=500:
        break

    folder = root / f"imgdata_{i}"
    folder.mkdir(parents=True, exist_ok=True)
    for name in ['im1','im2','im3']:
        imgs[name].save(folder/f"{name}.png")
```

Dataset is prepared for training. Now we have to check the exact format for training the Unet architecture for optical flow estimation and final flow correction.
